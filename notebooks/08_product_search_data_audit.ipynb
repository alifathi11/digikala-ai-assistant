# 08 — Product Search Data Audit

این نوت‌بوک قبل از طراحی Product Search، کیفیت و ساختار واقعی
`products_clean.parquet` و ارتباط آن با نظرات را بررسی می‌کند.

منطق audit در `src/rag/product_search/audit.py` قرار دارد و
notebook فقط orchestration و visualization است.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.rag.product_search import audit_product_data
```

```python
PRODUCTS_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "products_clean.parquet"
)

COMMENTS_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "comments_clean.parquet"
)

REPORT_PATH = (
    PROJECT_ROOT
    / "data"
    / "evaluation"
    / "product_search_data_audit.json"
)

print("Products:", PRODUCTS_PATH)
print("Comments:", COMMENTS_PATH)
```

```python
audit = audit_product_data(
    products_path=PRODUCTS_PATH,
    comments_path=COMMENTS_PATH,
    comment_batch_size=250_000,
    top_n=20,
)

audit.summary
```
