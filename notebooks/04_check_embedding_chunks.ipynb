# 04_check_embedding_chunks.ipynb

Check generated FAISS chunks and metadata.
No merge is performed.


```python
from pathlib import Path
import glob

import faiss
import pandas as pd
```

```python
INDEX_DIR = Path(
    "../data/indexes/product_comments_embedding"
)

CHUNK_DIR = INDEX_DIR / "chunks"

METADATA_PATH = INDEX_DIR / "metadata.parquet"
```

```python
chunk_files = sorted(
    glob.glob(
        str(CHUNK_DIR / "*.faiss")
    )
)

print("Number of chunks:", len(chunk_files))
print("First chunk:", chunk_files[0])
```

```python
total_vectors = 0

for file in chunk_files:

    index = faiss.read_index(file)

    print(
        Path(file).name,
        "vectors:",
        index.ntotal,
        "dimension:",
        index.d
    )

    total_vectors += index.ntotal


print("\nTotal vectors:", total_vectors)
```

```python
expected_count = 0

for file in chunk_files:
    index = faiss.read_index(file)
    expected_count += index.ntotal


print("Total vectors in chunks:", expected_count)
```
