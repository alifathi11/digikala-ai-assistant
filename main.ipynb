```python
import os, sys
import pandas as pd

os.environ.setdefault("MOCK_MODE", "true")

BASE_DIR = os.getcwd()
PRODUCTS_CSV = os.path.join(BASE_DIR, "data", "raw", "digikala-products.csv")
COMMENTS_CSV = os.path.join(BASE_DIR, "data", "raw", "digikala-comments.csv")

products_df = pd.read_csv(PRODUCTS_CSV, low_memory=False)
comments_df = pd.read_csv(COMMENTS_CSV, low_memory=False)

products_df = products_df.rename(columns={
    "title_fa": "title",
    "Rate":     "rate",
    "Brand":    "brand",
    "Category1":"category",
    "Price":    "price",
})

print(products_df.shape, comments_df.shape)
```

```python
from src.product_search import ProductSearcher
from src.comparator import compare_products
from src.category_analytics import low_recommendation_products, brand_comparison
```

```python
searcher = ProductSearcher(products_df)
results = searcher.search("هدفون بیسیم")
results = searcher.filter(results, max_price=5_000_000, min_rate=3.5)
results[["id", "title", "price", "rate"]].head()
```

```python
if len(results) < 2:
    print("کمتر از ۲ محصول یافت شد؛ مقایسه ممکن نیست.")
else:
    pid1 = int(results.iloc[0]["id"])
    pid2 = int(results.iloc[1]["id"])

    comparison = compare_products(
        product_ids=[pid1, pid2],
        query="کیفیت صدا چطوره؟",
        products_df=products_df,
    )

    for pid, data in comparison.items():
        print(f"\n{data['title']}:\n{data['answer']}")
```

```python
low_rec = low_recommendation_products(comments_df, products_df)
print("\nمحصولات با توصیه پایین:")
low_rec.head(10)

brand_stats = brand_comparison(comments_df, products_df, category="موبایل")
print("\nمقایسه برندها در دسته موبایل:")
brand_stats
```
